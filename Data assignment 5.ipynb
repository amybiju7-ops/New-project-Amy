## 6. Derived indicators
- **Sex ratio** = females per 1000 males
- **Literacy rate (%)** = literates / total population * 100
- **Work participation rate (%)** = total workers / total population * 100

#This code filters the data down to one row per state (combined rural+urban totals), calculates three indicators for each state — sex ratio, literacy rate, and work participation rate — then sorts states by population and shows the top 10 largest states along with these indicators.

```python
states = df[(df['Level'] == 'STATE') & (df['TRU'] == 'Total')].copy().reset_index(drop=True)

states['Sex Ratio (F per 1000 M)'] = (
    states['Total Population Female'] / states['Total Population Male'] * 1000
).round(1)
states['Literacy Rate (%)'] = (
    states['Literates Population Person'] / states['Total Population Person'] * 100
).round(2)
states['Work Participation Rate (%)'] = (
    states['Total Worker Population Person'] / states['Total Population Person'] * 100
).round(2)

states.sort_values('Total Population Person', ascending=False).head(10)

```

```python
#Top and bottom states by key indicators
This code prints four short rankings: the 5 most populous states, the 5 states with the highest literacy rate, the 5 states with the lowest literacy rate, and the 5 states with the highest sex ratio — each shown with just the state name and the relevant value.
```

```python
print("Top 5 most populous states:")
display(states.nlargest(5, 'Total Population Person')[['Name', 'Total Population Person']])

print("\nTop 5 states by literacy rate:")
display(states.nlargest(5, 'Literacy Rate (%)')[['Name', 'Literacy Rate (%)']])

print("\nBottom 5 states by literacy rate:")
display(states.nsmallest(5, 'Literacy Rate (%)')[['Name', 'Literacy Rate (%)']])

print("\nTop 5 states by sex ratio:")
display(states.nlargest(5, 'Sex Ratio (F per 1000 M)')[['Name', 'Sex Ratio (F per 1000 M)']])
```

#SC/ST population share

```python
states['SC Population (%)'] = (states['Scheduled Castes population Person'] / states['Total Population Person'] * 100).round(2)
states['ST Population (%)'] = (states['Scheduled Tribes population Person'] / states['Total Population Person'] * 100).round(2)
states.nlargest(5, 'ST Population (%)')[['Name', 'ST Population (%)']]
```

#Gender gap in literacy

```python
states['Male Literacy (%)'] = (states['Literates Population Male'] / states['Total Population Male'] * 100).round(2)
states['Female Literacy (%)'] = (states['Literates Population Female'] / states['Total Population Female'] * 100).round(2)
states['Literacy Gender Gap'] = (states['Male Literacy (%)'] - states['Female Literacy (%)']).round(2)
states.nlargest(5, 'Literacy Gender Gap')[['Name', 'Male Literacy (%)', 'Female Literacy (%)', 'Literacy Gender Gap']]
```

#Household size (avg persons per household)

```python
states['Avg Household Size'] = (states['Total Population Person'] / states['No of Households']).round(2)
states.nlargest(5, 'Avg Household Size')[['Name', 'Avg Household Size']]
```

#Child population share (0–6 age group) as a proxy for future demographic trends

```python
states['Child Population Share (%)'] = (states['Population in the age group 0-6 Person'] / states['Total Population Person'] * 100).round(2)
states.nlargest(5, 'Child Population Share (%)')[['Name', 'Child Population Share (%)']]
```

#Regional grouping (North/South/East/West/Northeast) using a mapping dict, then group-by comparisons

```python
region_map = {
    'JAMMU & KASHMIR': 'North', 'HIMACHAL PRADESH': 'North', 'PUNJAB': 'North',
    'CHANDIGARH': 'North', 'UTTARAKHAND': 'North', 'HARYANA': 'North',
    'NCT OF DELHI': 'North', 'RAJASTHAN': 'North', 'UTTAR PRADESH': 'North',

    'BIHAR': 'East', 'JHARKHAND': 'East', 'ODISHA': 'East', 'WEST BENGAL': 'East',

    'ASSAM': 'Northeast', 'ARUNACHAL PRADESH': 'Northeast', 'MANIPUR': 'Northeast',
    'MEGHALAYA': 'Northeast', 'MIZORAM': 'Northeast', 'NAGALAND': 'Northeast',
    'SIKKIM': 'Northeast', 'TRIPURA': 'Northeast',

    'MADHYA PRADESH': 'Central', 'CHHATTISGARH': 'Central',

    'GUJARAT': 'West', 'MAHARASHTRA': 'West', 'GOA': 'West',
    'DADRA & NAGAR HAVELI': 'West', 'DAMAN & DIU': 'West',

    'ANDHRA PRADESH': 'South', 'KARNATAKA': 'South', 'KERALA': 'South',
    'TAMIL NADU': 'South', 'PUDUCHERRY': 'South',
    'ANDAMAN & NICOBAR ISLANDS': 'South', 'LAKSHADWEEP': 'South'
}

states['Region'] = states['Name'].map(region_map)
states.groupby('Region')[['Literacy Rate (%)', 'Sex Ratio (F per 1000 M)']].mean().round(2)
```

## Story Pitch: "The Kerala Outlier"

**The hook:** In a country where men outnumber women in almost every major state, one state flips the script entirely.

**The story:** Kerala is the only major Indian state where women outnumber men (1,084 females per every 1,000 males), and it also tops the country in literacy at 84.2% — nearly 34 percentage points ahead of Bihar, the lowest-ranked major state (50.4%). This isn't a random data point; decades of research point to Kerala's investment in female education, healthcare access, and social welfare as the reason both numbers moved together.
 As India debates its next Census and updates to social welfare policy, Kerala's decades-old outcomes offer a real, measurable benchmark for what's achievable.
